```python
import os

class Student:
    """Represents an individual student record."""
    def __init__(self, student_id, name, age, grade):
        self.student_id = student_id
        self.name = name
        self.age = age
        self.grade = grade

    def to_string(self):
        """Converts object details to a comma-separated string for file storage."""
        return f"{self.student_id},{self.name},{self.age},{self.grade}\n"

    @staticmethod
    def from_string(record_str):
        """Creates a Student object from a comma-separated string."""
        data = record_str.strip().split(',')
        return Student(data[0], data[1], data[2], data[3])


class StudentManagementSystem:
    """Manages CRUD operations using file handling."""
    def __init__(self, filename="students.txt"):
        self.filename = filename
        # Ensure the file exists
        if not os.path.exists(self.filename):
            with open(self.filename, 'w') as f:
                pass

    def add_student(self):
        print("\n--- Add Student ---")
        student_id = input("Enter Student ID: ").strip()
        
        # Check if ID already exists
        if self._find_record(student_id):
            print("Error: A student with this ID already exists.")
            return

        name = input("Enter Name: ").strip()
        age = input("Enter Age: ").strip()
        grade = input("Enter Grade: ").strip()

        if not student_id or not name or not age or not grade:
            print("Error: All fields are required.")
            return

        student = Student(student_id, name, age, grade)
        try:
            with open(self.filename, 'a') as f:
                f.write(student.to_string())
            print("Student added successfully!")
        except IOError as e:
            print(f"File Error: Could not write data. {e}")

    def view_all_students(self):
        print("\n--- Student Records ---")
        try:
            with open(self.filename, 'r') as f:
                records = f.readlines()
            
            if not records:
                print("No records found.")
                return

            print(f"{'ID':<10}{'Name':<20}{'Age':<10}{'Grade':<10}")
            print("-" * 50)
            for record in records:
                s = Student.from_string(record)
                print(f"{s.student_id:<10}{s.name:<20}{s.age:<10}{s.grade:<10}")
        except IOError as e:
            print(f"File Error: Could not read data. {e}")

    def search_student(self):
        print("\n--- Search Student ---")
        student_id = input("Enter Student ID to search: ").strip()
        student = self._find_record(student_id)
        
        if student:
            print(f"\nRecord Found:\nID: {student.student_id}\nName: {student.name}\nAge: {student.age}\nGrade: {student.grade}")
        else:
            print("Student not found.")

    def update_student(self):
        print("\n--- Update Student ---")
        student_id = input("Enter Student ID to update: ").strip()
        student = self._find_record(student_id)
        
        if not student:
            print("Student not found.")
            return

        print(f"Current Details -> Name: {student.name}, Age: {student.age}, Grade: {student.grade}")
        new_name = input("Enter new Name (leave blank to keep current): ").strip()
        new_age = input("Enter new Age (leave blank to keep current): ").strip()
        new_grade = input("Enter new Grade (leave blank to keep current): ").strip()

        if new_name: student.name = new_name
        if new_age: student.age = new_age
        if new_grade: student.grade = new_grade

        self._update_file_record(student_id, student)
        print("Student record updated successfully!")

    def delete_student(self):
        print("\n--- Delete Student ---")
        student_id = input("Enter Student ID to delete: ").strip()
        if not self._find_record(student_id):
            print("Student not found.")
            return

        self._update_file_record(student_id, delete=True)
        print("Student record deleted successfully!")

    # Helper methods for internal operations
    def _find_record(self, student_id):
        try:
            with open(self.filename, 'r') as f:
                for line in f:
                    s = Student.from_string(line)
                    if s.student_id == student_id:
                        return s
        except IOError:
            pass
        return None

    def _update_file_record(self, student_id, updated_student=None, delete=False):
        """Handles both update and delete by rewriting the file without the targeted record."""
        try:
            with open(self.filename, 'r') as f:
                lines = f.readlines()

            with open(self.filename, 'w') as f:
                for line in lines:
                    s = Student.from_string(line)
                    if s.student_id == student_id:
                        if delete:
                            continue  # Skip writing this line to delete it
                        else:
                            f.write(updated_student.to_string())  # Write updated data
                    else:
                        f.write(line)  # Keep original data
        except IOError as e:
            print(f"File modification error: {e}")


def main():
    sms = StudentManagementSystem()
    while True:
        print("\n==============================")
        print("  STUDENT MANAGEMENT SYSTEM   ")
        print("==============================")
        print("1. Add Student")
        print("2. View All Students")
        print("3. Search Student")
        print("4. Update Student")
        print("5. Delete Student")
        print("6. Exit")
        
        choice = input("Select an option (1-6): ").strip()
        
        if choice == '1': sms.add_student()
        elif choice == '2': sms.view_all_students()
        elif choice == '3': sms.search_student()
        elif choice == '4': sms.update_student()
        elif choice == '5': sms.delete_student()
        elif choice == '6':
            print("Exiting system. Goodbye!")
            break
        else:
            print("Invalid choice. Please enter a number between 1 and 6.")

if __name__ == "__main__":
    main()
```
